## ¿Cómo selecciono columnas específicas de un DataFrame?

Estoy interesado en la edad de los pasajeros del Titanic.



```python
ages = titanic["Age"] # Edades de los pasajeros
ages.head()
```

Para seleccionar una sola columna, usa corchetes [] con el nombre de la columna de interés.

Cada columna en un DataFrame es una Serie. Cuando se selecciona una sola columna, el objeto devuelto es una Serie de pandas. Podemos verificar esto comprobando el tipo de la salida:

```python
type(titanic["Age"]) # Verifica que es una Serie de pandas
```

Y observemos la forma de la salida:

```python
titanic["Age"].shape # Devuelve (número de filas,)
```

DataFrame.shape es un atributo de una Serie y DataFrame de pandas que contiene el número de filas y columnas: (nfilas, ncolumnas). Una Serie de pandas es unidimensional y solo se devuelve el número de filas.

```python
age_sex.head() # Muestra las primeras filas del nuevo DataFrame
```

Para seleccionar múltiples columnas, usa una lista de nombres de columna dentro de los corchetes de selección [].

```python
type(titanic[["Age", "Sex"]]) # Verifica que es un DataFrame
```

## ¿Cómo filtro filas específicas de un DataFrame?

Estoy interesado en los pasajeros mayores de 35 años.



```python
above_35 = titanic[titanic["Age"] > 35] # Pasajeros con edad mayor a 35
above_35.head() # Muestra las primeras filas
```

Para seleccionar filas basándose en una expresión condicional, usa una condición dentro de los corchetes de selección [].

La condición dentro de los corchetes de selección titanic["Age"] > 35 verifica en qué filas la columna "Age" tiene un valor mayor a 35:



```python
titanic["Age"] > 35 # Serie booleana: True si la edad es > 35
```

La salida de la expresión condicional (> pero también ==, !=, <, <=,… funcionarían) es en realidad una Serie de pandas de valores booleanos (True o False) con el mismo número de filas que el DataFrame original. Tal Serie de valores booleanos se puede usar para filtrar el DataFrame colocándola dentro de los corchetes de selección []. Solo se seleccionarán las filas para las que el valor sea True.

Sabemos de antes que el DataFrame original del Titanic consta de 891 filas. Observemos el número de filas que cumplen la condición verificando el atributo shape del DataFrame resultante above_35:

```python
above_35.shape # Muestra (n_filas_cumplen, n_columnas)
```

Estoy interesado en los pasajeros del Titanic de las clases 2 y 3.

```python
class_23 = titanic[titanic["Pclass"].isin([2, 3])] # Filtra por clase 2 o 3
class_23.head() # Muestra las primeras filas
```

Similar a la expresión condicional, la función condicional isin() devuelve True para cada fila cuyos valores están en la lista proporcionada. Para filtrar las filas basándose en dicha función, usa la función condicional dentro de los corchetes de selección []. En este caso, la condición dentro de los corchetes titanic["Pclass"].isin([2, 3]) verifica en qué filas la columna "Pclass" es 2 o 3.

Lo anterior es equivalente a filtrar por filas cuya clase es 2 o 3 y combinar las dos declaraciones con un operador | (o):

```python
class_23 = titanic[(titanic["Pclass"] == 2) | (titanic["Pclass"] == 3)]
```

```python
class_23.head() # Muestra las primeras filas
```

Nota:

Al combinar múltiples declaraciones condicionales, cada condición debe estar entre paréntesis (). Además, no se puede usar "or" o "and" sino que se deben usar el operador | para "o" y el operador & para "y".



Quiero trabajar con los datos de pasajeros cuya edad se conoce.

```python
age_no_na = titanic[titanic["Age"].notna()] # Filtra filas donde edad no es nula
age_no_na.head() # Muestra las primeras filas
```

La función condicional notna() devuelve True para cada fila cuyos valores no son nulos. Así, esto se puede combinar con los corchetes de selección [] para filtrar la tabla de datos.

Te preguntarás qué cambió realmente, ya que las primeras 5 líneas siguen siendo los mismos valores. Una forma de verificar es comprobar si la forma (shape) ha cambiado:

```python
age_no_na.shape # Muestra (n_filas_no_nulas, n_columnas)
```

## ¿Cómo selecciono filas y columnas específicas de un DataFrame?

Estoy interesado en los nombres de los pasajeros mayores de 35 años.



```python
adult_names = titanic.loc[titanic["Age"] > 35, "Name"] # Nombres de pasajeros > 35 años
adult_names.head() # Muestra los primeros nombres
```

En este caso, se hace un subconjunto tanto de filas como de columnas de una sola vez y solo usar corchetes de selección [] ya no es suficiente. Se requieren los operadores loc/iloc delante de los corchetes []. Al usar loc/iloc, la parte antes de la coma son las filas que deseas, y la parte después de la coma son las columnas que deseas seleccionar.

Cuando uses nombres de columna, etiquetas de fila o una expresión condicional, usa el operador loc delante de los corchetes de selección []. Tanto para la parte antes como después de la coma, puedes usar una sola etiqueta, una lista de etiquetas, un segmento (slice) de etiquetas, una expresión condicional o dos puntos. Usar dos puntos especifica que deseas seleccionar todas las filas o columnas.

Estoy interesado en las filas 10 a 25 y las columnas 3 a 5.

```python
titanic.iloc[9:25, 2:5] # Selecciona filas 10-25 (índice 9 a 24) y columnas 3 a 5 (índice 2 a 4)
```

Nuevamente, se hace un subconjunto tanto de filas como de columnas de una sola vez y solo usar corchetes [] ya no es suficiente. Cuando estés específicamente interesado en ciertas filas y/o columnas basándote en su posición en la tabla, usa el operador iloc delante de los corchetes [].

Al seleccionar filas y/o columnas específicas con loc o iloc, se pueden asignar nuevos valores a los datos seleccionados. Por ejemplo, para asignar el nombre "anonymous" a los primeros 3 elementos de la cuarta columna:

```python
titanic.iloc[0:3, 3] = "anonymous" # Asigna "anonymous" a las primeras 3 filas de la columna 4
titanic.iloc[:5, 3] # Muestra las primeras 5 filas de la columna 4
```
