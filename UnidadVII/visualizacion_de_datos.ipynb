## Visualización de datos


A continuación, revisaremos una serie de gráficos útiles que nos ayudarán a familiarizarnos con los datos y a extraer información clave de manera visual.

Usaremos el dataset penguins el cual contiene mediciones corporales de pingüinos en tres islas del Archipiélago Palmer.

A continuación cargaremos el conjunto de datos penguins usando la libreria Pandas.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


url = "https://raw.githubusercontent.com/bayesianoues/intro_progra_biologia_2026/refs/heads/main/Datasets/penguins.csv"
penguins = pd.read_csv(url)

```

```python
penguins.head()
```

## Gráficos Univariados

## Gráfico de Barras

```python
# Gráfico de barras con seaborn
sns.countplot(data=penguins, x='species')
plt.title('Conteo de pingüinos por especie')
plt.xlabel('Especie')
plt.ylabel('Cantidad')
plt.show()
```

El gráfico muestra la distribución del número de pingüinos por especie, donde se observa claramente que la especie Adelie es la más abundante, con una cantidad significativamente mayor en comparación con las demás. En segundo lugar se encuentra Gentoo, con una presencia también considerable pero menor que Adelie, mientras que Chinstrap es la especie menos representada en el conjunto de datos.

## Histograma

```python
# Histograma básico
sns.histplot(data=penguins, x='body_mass_g')
plt.title('Distribución de la masa corporal de pingüinos')
plt.xlabel('Masa corporal (gramos)')
plt.ylabel('Frecuencia')
plt.show()
```

Cuando se construye un histograma es necesario ajustar el número de intervalos de tal forma que refleje la forma de los datos, esto esta fuera del alcance del curso, sin embargo, esto se tendria que presentar en los cursos de estadística.

A efectos ilustrativos se comenta el histograma.

El histograma muestra la distribución de la masa corporal de los pingüinos, evidenciando una concentración mayor de individuos en el rango aproximado entre 3500 y 4500 gramos, donde se observa el pico de frecuencia. A medida que la masa corporal aumenta hacia valores superiores (por encima de 5000 gramos), la frecuencia disminuye progresivamente, lo que sugiere que los pingüinos más pesados son menos comunes en el conjunto de datos. Asimismo, existe una ligera asimetría hacia la derecha (cola larga hacia masas mayores), indicando la presencia de algunos individuos significativamente más pesados, pero en menor proporción. En conjunto, la distribución permite inferir que la mayoría de los pingüinos tienen una masa corporal intermedia, con pocos casos extremos en ambos extremos del rango.

## Gráficos Bivariados

## Dos variables numéricas

## Gráfico de dispersión

```python
# Gráfico de dispersión básico
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g')
plt.title('Relación entre longitud de aleta y masa corporal')
plt.xlabel('Longitud de aleta (mm)')
plt.ylabel('Masa corporal (g)')
plt.show()
```

El diagrama de dispersión muestra una relación positiva clara entre la longitud de la aleta y la masa corporal de los pingüinos: a medida que aumenta la longitud de la aleta, también tiende a incrementarse la masa corporal. Los puntos siguen una tendencia ascendente bastante definida, lo que sugiere una correlación fuerte entre ambas variables. Además, se pueden observar posibles agrupamientos de datos, lo que podría indicar la presencia de diferentes especies con características físicas distintas. Aunque existe cierta variabilidad, especialmente en valores intermedios, el patrón general confirma que los pingüinos con aletas más largas suelen ser más pesados.

## Dos variables categóricas

## Gráfico de barras apilado

```python
# Barras apiladas con conteos

sns.histplot(data=penguins,
             x='island',
             hue='species',
             multiple='stack',
             shrink=0.8)

plt.title('Conteo de especies por isla (barras apiladas)')
plt.xlabel('Isla')
plt.ylabel('Conteo')

plt.show()
```

El gráfico de barras apiladas muestra la distribución de las especies de pingüinos en las distintas islas, evidenciando diferencias claras en su presencia. En Torgersen, únicamente se observa la especie Adelie, lo que sugiere una distribución exclusiva en esa isla. Por otro lado, Biscoe presenta la mayor cantidad total de pingüinos, dominada principalmente por la especie Gentoo, aunque también incluye Adelie en menor proporción. Finalmente, en Dream se observa una combinación de especies, donde Chinstrap tiene una presencia importante junto con Adelie. En conjunto, el gráfico permite identificar cómo la distribución de las especies varía significativamente según la isla, lo que podría estar relacionado con factores ambientales o ecológicos específicos de cada ubicación.

## Una variable numérica y una variable categórica

```python
# Boxplot básico
sns.boxplot(data=penguins, x='species', y='body_mass_g')

plt.title('Distribución de masa corporal por especie')
plt.xlabel('Especie')
plt.ylabel('Masa corporal (g)')
plt.show()
```

El diagrama de cajas muestra la distribución de la masa corporal de los pingüinos según su especie, permitiendo comparar tanto la tendencia central como la variabilidad. Se observa que la especie Gentoo presenta las masas corporales más altas, con una mediana claramente superior y una mayor dispersión en comparación con las otras especies. Por otro lado, Adelie y Chinstrap tienen distribuciones más similares entre sí, con medianas más bajas y menor variabilidad, aunque Chinstrap muestra algunos valores atípicos tanto en el extremo inferior como superior. En conjunto, el gráfico evidencia diferencias significativas entre especies, sugiriendo que la masa corporal es una característica distintiva importante entre ellas.

## Subplots

Los subplots nos permiten incluir otra variable categorico en los gráficos. Aqui va un ejemplo.

```python
g = sns.FacetGrid(penguins, col='island', height=4, aspect=1.2)
g.map(sns.scatterplot, 'flipper_length_mm', 'body_mass_g', 'species')
g.add_legend()

plt.show()
```

El conjunto de subgráficos muestra la relación entre la longitud de la aleta y la masa corporal de los pingüinos, segmentada por isla, lo que permite comparar patrones entre ubicaciones. En Torgersen, se observa principalmente la especie Adelie, con una correlación positiva moderada entre ambas variables y una dispersión relativamente acotada. En Biscoe, aparecen dos grupos claramente diferenciados: uno correspondiente a Adelie (valores más bajos) y otro a Gentoo, que presenta mayores longitudes de aleta y masas corporales, evidenciando una separación marcada entre especies. Por último, en Dream se observa una mezcla de Adelie y Chinstrap, con cierta superposición pero manteniendo la tendencia positiva general. En conjunto, los subplots muestran que la relación entre longitud de aleta y masa corporal es consistente en todas las islas, aunque la composición de especies introduce diferencias claras en los rangos y agrupamientos observados.

Para ilustrar el tema de los mapas de gradientes calor vamos a usar el conjunto de datos de vino el cual se ha descargado del sitio web  UC Irvine Machine Learning Repository.

[Información del conjunto de datos de vinos](https://github.com/bayesianoues/intro_progra_biologia_2026/blob/main/Datasets/winequality_descripcion.md)

El mapa de gradiente de calor puede ser útil para mostrar la estructura de correlación de las variables cuantitativas de un conjunto de datos, a continuación se muestra esto con el conjunto de datos de vino rojo.


```python

# Cargar el conjunto de datos desde UCI
url = "https://raw.githubusercontent.com/bayesianoues/intro_progra_biologia_2026/refs/heads/main/Datasets/winequality_red.csv"
wine_red = pd.read_csv(url, sep=';')

# Inspeccionar los datos
wine_red.head()
```

```python
# Calcular la matriz de correlación
corr = wine_red.corr()

# Establecer el estilo de seaborn
sns.set(style="white")

# Crear el mapa de calor con un mapa de gradiente
plt.figure(figsize=(12, 8))
heatmap = sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",   # Mapa de gradiente
    center=0,          # Centrar alrededor de cero
    linewidths=0.5
)

plt.title("Mapa de calor de correlación con mapa de gradiente (Conjunto de datos de calidad del vino)")
plt.show()
```

En el anterior gráfico se muestra lo siguiente:

* Correlaciones positivas fuertes (por ejemplo, alcohol vs. calidad) → rojo

* Correlaciones negativas fuertes → azul

* Correlación débil o nula → cerca de blanco

Aquí es exactamente donde los mapas de gradiente destacan: codifican visualmente la estructura continua en los datos.

## Gráficos en 3D

Como ejemplo de gráfico en 3D realizaremos un gráfico de dispersión con los datos del dataset iris.

El gráfico consitira en los siguientes ejes y además incorporaremos el color a los puntos.

* X → sepal length

* Y → sepal width

* Z → petal length

* Color → species

En primer lugar cargaremos el dataset con el siguiente código:

```python
filename = 'https://raw.githubusercontent.com/bayesianoues/intro_progra_biologia_2026/refs/heads/main/Datasets/iris.csv'
names = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'class']

iris = pd.read_csv(filename, names=names)

iris.head()
```

A continuación realizaremos el gráfico de dispersión en 3D.

```python
from mpl_toolkits.mplot3d import Axes3D

# Establecer el estilo de seaborn
sns.set(style="whitegrid")

# Crear figura 3D
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

# Mapear especies a colores
species = iris['class'].unique()
colors = ['red', 'green', 'blue']

for sp, color in zip(species, colors):
    subset = iris[iris['class'] == sp]
    ax.scatter(
        subset['sepal-length'],
        subset['sepal-width'],
        subset['petal-length'],
        label=sp,
        color=color,
        s=50
    )

# Etiquetas
ax.set_xlabel("Longitud del sépalo")
ax.set_ylabel("Ancho del sépalo")
ax.set_zlabel("Longitud del pétalo")

# Título y leyenda
ax.set_title("Gráfico de dispersión 3D - Conjunto de datos Iris")
ax.legend()

plt.show()
```

El gráfico anterior muestra una dispersión tridimensional del conjunto de datos Iris utilizando tres variables, donde cada color representa una especie distinta. Se observa que la clase Iris-setosa (rojo) forma un grupo claramente separado del resto, lo que indica que sus características son fácilmente distinguibles en este espacio. En cambio, Iris-versicolor (verde) y Iris-virginica (azul) presentan cierto grado de solapamiento, aunque aún se percibe una tendencia de separación, especialmente a medida que aumenta la tercera dimensión. En conjunto, el gráfico sugiere que algunas especies son linealmente separables mientras que otras requieren modelos más complejos para su correcta clasificación.
