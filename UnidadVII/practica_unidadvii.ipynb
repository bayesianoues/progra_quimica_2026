1. Leer los datos

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


filename = 'https://raw.githubusercontent.com/bayesianoues/intro_progra_biologia_2026/refs/heads/main/Datasets/playground_series_mar_2021.csv'

dat = pd.read_csv(filename)
```

4. Usando .iloc recupere la fila cuyo indice esta en la quinta posición.

5. Cuantas filas cumplen con los siguientes criterios:

* cat1 es igual a A y cat2 es igual a A

* Solo recupere las columnas id, cat1, cat2 y target

*Sugerencia*: Use el método head para mostrar los resultados.

```python
filtered_dat = dat.loc[(dat["cat1"] == "A") & (dat["cat2"] == "A" ), ["id", "cat1", "cat2", "target"]]
```

```python
print(filtered_dat.shape)
```

```python
filtered_dat.head()
```

6. Resumir el número de valores únicos en cada columna

```python
# Resumir el número de valores únicos en cada columna
for i in range(dat.shape[1]):
  print(i, len(np.unique(dat.iloc[:, i])))
```

En caso de existir variables que solo tiene un valor, eliminarlas.

7. Contar el número de valores ausentes por columnas.

```python
dat.isnull().sum()
```

8. Usando una variable categorica construya un grafico de barras.

```python
# Gráfico de barras
sns.countplot(data=dat, x='cat1')
plt.title('Conteo de la variable cat1')
plt.xlabel('Cat1')
plt.ylabel('Cantidad')
plt.show()
```

9. Usando una variable cuantitativa construya un histograma.

```python
# Histograma básico
sns.histplot(data=dat, x='cont1')
plt.title('Distribución de la variable cont1')
plt.xlabel('Cont1')
plt.ylabel('Frecuencia')
plt.show()
```

10. Construya un gráfico de dispersión entre dos variables cuantitativas.

```python
# Gráfico de dispersión básico
sns.scatterplot(data=dat, x='cont1', y='cont2')
plt.title('Relación entre las variables cont1 y cont2')
plt.xlabel('Cont1')
plt.ylabel('Cont2')
plt.show()
```
