### Descripción del conjunto de datos

Este conjunto de datos está relacionado con las variantes de vino tinto del vino portugués «Vinho Verde». El conjunto de datos describe las cantidades de diversas sustancias químicas presentes en el vino y su efecto sobre la calidad del mismo.

El conjunto de datos puede abordarse como un problema de clasificación o de regresión. Las clases están ordenadas y no están balanceadas (por ejemplo, hay muchos más vinos de calidad normal que vinos excelentes o deficientes).

**Este conjunto de datos contiene las siguientes variables:**

**Variables de entrada (basadas en pruebas fisicoquímicas):**

1. Acidez fija (*fixed acidity*).
2. Acidez volátil (*volatile acidity*).
3. Ácido cítrico (*citric acid*).
4. Azúcar residual (*residual sugar*).
5. Cloruros (*chlorides*).
6. Dióxido de azufre libre (*free sulfur dioxide*).
7. Dióxido de azufre total (*total sulfur dioxide*).
8. Densidad (*density*).
9. pH.
10. Sulfatos (*sulphates*).
11. Alcohol (*alcohol*).

**Variable de salida (basada en datos sensoriales):**

12. Calidad (*quality*): puntuación entre 0 y 10.
